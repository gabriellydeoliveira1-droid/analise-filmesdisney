Informações das variaveis:

Quantidades de Colunas: 6

movie_title Object - Titulo ou Filme ( ex: Snow White, Piocchio, Fantasia, Dalmatas, CInderela e etc.)

release_date Object - Ano do Filme ( ex: 21/12/1937 )

genre Object - Genero do Filme ( ex: Musical, Drama, Comedia e etc.)

mpaa_rating Object - Classificação indicada (ex: G significa para todas as idades, PG-13 Não recomendado para menores de 13.)

total_gross Int - Total de vendas de bilheteria (ex: Quanto dinheiro foi arrecadado)

inflation_adjusted_gross - Int - Arrecadação ajustada pela infração (ex: Ela mostra quanto o filme arrecadaria nos valores atuais do dinheiro.)

```python
import pandas as pd #pandas
import matplotlib.pyplot as plt #matplots
import seaborn as sns #seaborn
#df = dataframes
df = pd.read_csv('disney_movies.csv' , sep = ',')
df.head() #5 primeiro dados da tabela
```

```python
df = df[['movie_title', 'genre']]
df.head(10)
```

```python
df.sort_values('movie_title', ascending=False).head(10)
```

```python
plt.figure(figsize=(10,6))

genre_counts = df['genre'].value_counts()

sns.barplot(
    x=genre_counts.values,
    y=genre_counts.index
)

plt.title('Gêneros Mais Frequentes')
plt.xlabel('Quantidade')
plt.ylabel('Gênero')

plt.show()
```

```python
plot_df = pd.read_csv('disney_movies.csv' , sep = ',')

top_filmes = plot_df.sort_values(
    'inflation_adjusted_gross',
    ascending=False
).head(10)

top_filmes['filme_genero'] = (
    top_filmes['movie_title']
    + ' (' +
    top_filmes['genre'] + ')'
)


plt.figure(figsize=(12,6))

sns.barplot(
    data=top_filmes,
    x='inflation_adjusted_gross',
    y='filme_genero'
)

plt.title('Filmes Mais Assistidos por Gênero')
plt.xlabel('Arrecadação Ajustada')
plt.ylabel('Filme e Gênero')

plt.show()
```
